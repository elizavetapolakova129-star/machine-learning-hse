**Проблематика**

Существует множество исследований, которые описывают взаимосвязь продолжительности жизни с различными факторами, но анализ проводится с фактом (или группой схожих по смыслу факторов) по отдельности, нет единой модели рассматривающей несколько сторон сразу.  Рассматривается, например, влияние только факторов риска, таких как употребление алкоголя и уровнь расспространенности различных заболеваний, в отрыве от других показателей, исследователи просо утверждают о отрицательной связи между продолжительностью жизни и факторами риска, также анализ строится на основе анализа корреляций, а не регресии, так что мы теряем возможность анализа изменения одной переменной в зависимости от изменения другой (Масленникова, Оганов, 2013). Более того, существуют также разные позиции касательно влияния некоторых факторов, так исследования говорят о сильной взаимосвязи ВВП и продолжительности жизни (Разводовский, Голенков, 2020; Lubis, 2024). Но в тоже время есть данные о том, что при высоком уровне ВВП это связь начинает ослабевать, что означет убывающую отдачу от роста ВВП (Владимирская, Колосницына, 2023).  

Самой главной проблемой является тот факт, что никто не рассматривает совокупное влияние социально-экономических и "медицинских" факторов на продолжительность жизни в единой модели, так что в этой работе я постараюсь рассмотреть совместное влияние разных аспектов

**Гипотезы**

Содержательные:

*Н1*: Употребление алкоголя сокрщает продолжительность жизни.  

Существует отрицательная связь между потреблением алкоголя и ожидаемой продолжительностью жизни как для мужчин, так и для женщин. Увеличение совокупного объема потребления алкогольных напитков оказывает негативное воздействие на здоровье населения, что приводит к формированию значительного различия в ожидаемой продолжительности жизни (Коссова и др., 2017).

*Н2*: Распространенность ВИЧ-инфекции сокращает продолжительность жизни.

Распространение вируса иммунодефицита человека (ВИЧ) привело к пандемии заболевания ВИЧ/СПИДом, повлиявшей на численность и продолжительность жизни населения. ВИЧ-инфекция является пожизненным заболеванием, которое через несколько лет приводит к развитию смертельно опасного синдрома приобретенного иммунного дефицита (СПИД). Среди смертей от всех причин доля смертей от ВИЧ/СПИДа достигает 1%, а в возрастной группе 30-39 лет превышает 5%. Наиболее поражена ВИЧ возрастная группа 35-39 лет, в которой ВИЧ-инфицированные составляют 2% населения, что свидетельствует о значительных потерях в наиболее трудоспособной части населения (Покровский и др., 2017).

*Н3*: Повышение уровня образования увеличивает продолжительность жизни.

Люди с более высоким образовательным статусом реже сталкиваются с сердечно-сосудистыми и желудочно-кишечными патологиями по сравнению со сверстниками со средним образованием. Кроме того, респонденты с высшим образованием чаще обращаются за профилактическими медицинскими осмотрами, что способствует раннему выявлению и предотвращению заболеваний. Более высокий образовательный статус способствует лучшему пониманию вопросов здоровья, более активному использованию медицинских ресурсов и формированию здорового образа жизни, что в совокупности приводит к увеличению средней продолжительности жизни.(Абдоков и др., 2025)

*Н4*: Проживание в развитой стране увеличивает продолжительность жизни.

Ожидаемая продолжительность жизни существенно различается между группами стран в зависимости от уровня экономического развития, расходов на здравоохранение и других социально-экономических факторов. В зависимости от принадлежности стран к разным кластерам по уровню продолжительности жизни на здоровье населения влияют различные факторы и в неодинаковой мере: развитие экономики, включая здравоохранение, урбанизация, питание, а также нездоровый образ жизни. Это подтверждает, что принадлежность страны к группе развитых или развивающихся является значимым предиктором ожидаемой продолжительности жизни (Колосницына, 2019).

*Н5*: Проживание в стране с высоким ВВП увеличинает продолжительность жизни.

Связь между реальным ВВП на душу населения и ожидаемой продолжительностью жизни является положительной и имеет умеренный характер, ВВП на душу населения является важным, хотя и не единственным фактором, влияющим на продолжительность жизни. Страны с высоким уровнем ВВП демонстрируют более высокие показатели продолжительности жизни, однако характер взаимосвязи различается в зависимости от принадлежности стран к группам с высоким, средним и низким уровнем развития (Беданоков и др., 2022).

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

Алкоголь - поведенческий фактор риска, который повышает вероятность развития сердечно-сосудистых заболеваний, цирроза печени, онкологии и травматизма, что в совокупности снижает здоровье населения.Измеряется как oбъемы потребления алкоголя в литрах среди людей старше 15 лет

ВИЧ/СПИД - отражает масштаб эпидемии, которая подрывает здоровье населения в трудоспособном возрасте, увеличивает смертность и создает нагрузку на систему здравоохранения. Измеряется как число смертей от СПИДа среди детей в возрасте до 4 лет на 1000 человек, что как раз отражает распространеность инфекции.

Уровень экономического развития опеределяет ресурсы, доступные для здравоохранения, образования и социальной защиты, что влияет на здоровье населения.Бинарная переменная, где 1 - развитая страна, 0 - развивающаяся

ВВП отражает материальные ресурсы страны, доступные для инвестиций в здравоохранение, инфраструктуру и социальные программы. Измеряется как ВВП на душу населения (в текущих долларах США)

Датасет включает все пременные, по которым составлены гипотезы. Сам датасет включает множество переменных, все эти переменные направлена на описание продолжительности жизни, есть как данные касающиеся медицинских аспектов, так и социально-экономические, что как раз и необходимо для меня исходя из выявленной проблематики. Данные отсортированы по странам в алфавитном порядке, в рамках одной страны от 2015 до 2000

```python
df = pd.read_csv('life_expectancy.csv')
df.sample(5)
```

```python
df = df[['Life expectancy', 'Alcohol', 'HIV/AIDS', 'Schooling', 'Status', 'GDP']]
df = df.dropna()
df
```

**Разведывательный анализ**

Для начала посмотрим на распределение наших непрерывных переменных(то есть всех, кроме уровня развития), автоматически выводит без этого столбца

```python
df.describe()
```

Теперь посмотрим распреление наших непрерывных величин с помощью гистограмм

```python
df.hist(figsize=(12, 8), color ='red', layout=(2, 3))
plt.suptitle('Гистограммы непрерывных переменных')
plt.show()
```

Теперь преобразуем наши переменные, во-первых, добавим логарфмирование для ВВП, тк финансовая переменная, а также для ВИЧ и уровня образования для соблюдения линейности, во-вторых, преобразуем уровень развития страны в дамми-переменную.

```python
df['log_GDP'] = np.log(df['GDP'])
df['log_HIV'] = np.log(df['HIV/AIDS'] + 0.1) # добавление 0.1 необходимо для обработки около 0 случаев (*без этого добавления не линейна), добавляем так мало, потому что значения сконцентрированы около 0 и будем чувствительны к сильным изменениям
df['log_Schooling'] = np.log(df['Schooling'] + 1)  # добавление 1 необходимо для обработки случая с 0 лет образования (log(0) ошибка, нельзя даже посторить модель), добавляем 1 тк мало значений около 0 и мы не так чувсвительны к ним (можно +0.1, но R^2 будет поменьше)
df['is_developed'] = np.where(df['Status'] == 'Developed', 1, 0)
```

```python
df = df.drop(['GDP', 'HIV/AIDS', 'Status', 'Schooling'], axis=1)
```

Теперь посмотрим на корреляцию наших переменных

```python
ax = sns.heatmap(df.corr(numeric_only=True), cmap="vlag", annot=True, fmt="0.2f")
```

*Наконец посторим множественную линейную регрессию*

**Обоснование спецификаций модели**

Были выбраны перменные потребления алкоголя, смертности от СПИДа, уровень образования, ВВП и уровень развития страны на основе теоретических данных использованных в обосновании гипотез. Переменная алкоголя включена линейно, тк согласно исследованиям существует линейная свзяь между увеличение дозы и ростом риска смертности (Kassaw et al., 2024). Переменная ВВП классическая финансовая переменная, так что вклюена в модель с логарифмическим преобразованием. Уровень развития страны явлется фиктивной переменной, которая контроирует институциональные различия меду странами. Переменные уровня образования и смертности от СПИДа также включена с логарифмическими перобразованиями для учета нелинейной связи эти переменных с продолжительностью жизни (Everett BG et al., 2013; Marcus JL et al., 2016). Также в модель осознанно не включены переменные года и страны, то есть я не контралирую этот эффект, еси бы они были включена то при выводе регрессии было бы более 200 строк, что очевидно тяжело интерпретировать

Мою было принято решение не удалять выбросы, ведь в изначальном датасете это реальные данных по годам, те в таком случае мы фактически можем игнорировать страны с очень хорошими и плохими показателями

```python
model = smf.ols('Q("Life expectancy") ~ Alcohol + log_HIV + is_developed + log_Schooling + log_GDP', data=df).fit()
print(model.summary())
```

**Проверка всех условий Гаусса-Маркова**

Для начала проверим линейность

```python
from statsmodels.stats.diagnostic import het_breuschpagan, normal_ad, linear_reset
reset_test = linear_reset(model, power=2, use_f=True)
print(reset_test)
if reset_test.pvalue > 0.05:
    print("Модель специфицирована корректно (нет пропущенных нелинейностей, p > 0.05)")
else:
    print("Модель может быть неверно специфицирована (p ≤ 0.05), возможно, нужны нелинейные члены")
```

Линейность соблюдена, теперь проверим мультиколлениарность, переменная развития бинарная, так что ее не включаем в проверку

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
X_vif = df[['Alcohol', 'log_HIV', 'log_Schooling', 'log_GDP']].copy()
X_vif = sm.add_constant(X_vif)
vif_data = pd.DataFrame({
    'Переменная': X_vif.columns,
    'VIF': [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])]
})
print(vif_data)
```

Мультиколлениарность нет, тк значения VIF для всех наших переменных (очевидно константу не считаем) меньше даже 5, перейдем к проврке на гомоскедастичность

```python
from statsmodels.stats.diagnostic import het_breuschpagan
bp_test = het_breuschpagan(model.resid, model.model.exog)
print(bp_test)
```

Второе число в выводе это p-value и оно близко к 0, а значит Н0 отвергается и у нас есть гетероскадастичность, воспользуемся методом робастных ошибок, а именно HC1 (выборка больше 2к наблюдений)

```python
robust_model = model.get_robustcov_results(cov_type='HC1')
print(robust_model.summary())
```

Проверим наши данные на автокорреляцию

```python
from statsmodels.stats.stattools import durbin_watson
durbin_watson(model.resid)
```

Значение теста далеко от 2, что говрить о сильной положтельной автокорреляции, что в целом не удивительно, ведь в изначальном датасете данных разделены по странам с наблюдениями за несколько лет, так что для такого датасета скорее было бы странно увидеть оствустиве автокорреляции

И последнее, проверим нормальность

```python
from scipy.stats import shapiro
residuals = model.resid
shapiro_stat, shapiro_p = shapiro(residuals)
print(f"\nТест Шапиро-Уилка: статистика = {shapiro_stat:.3f}, p-value = {shapiro_p:.4f}")
if shapiro_p > 0.05:
    print("Остатки распределены нормально (p > 0.05)")
else:
    print("Остатки не подчиняются нормальному распределению (p ≤ 0.05)")
```

Для наглядности нарисует Q-Q плот

```python
plt.figure(figsize=(5,4))
stats.probplot(residuals, dist="norm", plot=plt)
plt.title('Q-Q plot остатков')
plt.show()
```

Хотя из теста остсатки не подчиняются нормально распределению, в целом, из графика можно сделать, что распределение все таки не является таким критически не нормальным, достаточно ожидаемая картина для правдивых данных. Учитывая также, что в начале было принято решение не очищать выбросы, то "тяжелые хвосты" тоже вполне ожидаемы

При проверке тестов было принято решение не применять метод Бокса-Кокса, ведь нам как раз важна интерпретация для выводов по гипотезам, так что итого остаются не соблюденными 2 условия: нормальность распреления остатков и автокорреляция

Еще хотелось бы заметить, что R^2 у модели более 80%, что очень хороший показатель)))

Теперь к содержательным гипотизам

Н1: Употребление алкоголя сокрщает продолжительность жизни.

Эта гипотеза не подтвердилась, согласно модели наоборот потребление алкоголя увеличинвает продолжительность жизни. Исследования показывают, что важна не столько абсолютное количество потребляемого алкоголя, сколько паттерн потребления. В странах с высокой продолжительностью жизни, таких как Франция или Италия распространена культура умеренного потребления алкоголя во время еды, тогда как в странах с низкой продолжительностью жизни алкоголь часто потребляется в форме запоев, что наносит значительно больший вред здоровью (Holahan, 2014). В модели используется только количественный показатель (литров на душу), который не отражает эти качественные различия. То есть фактически мы скорее говорим о пользе от употребления небольшого количества, ведь давно известно, что это помогает работе сердца, а также работе мозга, ведь помогает человеку избавится от стресса и лучше спать (Пальман, Полуэктов, 2022).

Н2: Распространенность ВИЧ-инфекции сокращает продолжительность жизни.

Эта гипотеза подвердилась и соотвествует указанной изначально литературе, все просто, расспространение инфекции ведет к тому, что многие люди попросту могут не знать о заболевании и не "лечить" его, а значит оно быстрее будет прогрессировать, перейдет в стадию СПИДа, человек заразится легкой простудой и умрет, тем самым общая продолжительность жизни будет сокращаться (Покровский и др., 2017).

Н3: Повышение уровня образования увеличивает продолжительность жизни.

Эта гипотеза также подтвердилась, что также согласуется с теоретическими данными, что тоже ожидаемо, ведь более образованные люди лучше понимают сигналы своего тела, чаще более отвественно относятся к личению, чаще ходят на профилакические осмотры (Абдоков и др., 2025).

Н4: Проживание в развитой стране увеличивает продолжительность жизни.

Эта гипотеза тоже подтвердилась. Дело в том, что развитые страны потому и развитые, что тратят много денег на обеспечение благополучия граждан, они заботятся о информированности населения о заболеваниях, постоянно улучшают систему здравоохранения, что согласуется с результатами из статьи  Колосницыной и др.

Н5: Проживание в стране с высоким ВВП увеличинает продолжительность жизни.

Эта гипотиза тоже подтвердилась, что также заметно подтвердилась и умеренная связь, о которой говорится в теории в статье Беданокова и др., на фоне всех остальных признаков ВВП имеет самое маленькое значение коэффицента, это связь в большей степени обусловена тем, что чаще высокий ВВП имеют развитые страны.
